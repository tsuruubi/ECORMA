```python
# === Settings: Output directory (Modify as needed) ===
# Define the path for exporting model data.
out_dir = current_dir.parent / "fba/output/model_exports"
out_dir.mkdir(parents=True, exist_ok=True)

def sanitize_for_filename(s: str) -> str:
    """Sanitizes strings for safe use in filenames by removing invalid characters."""
    return re.sub(r'[^0-9A-Za-z._-]+', '_', str(s))[:120]

def extract_objective_info(model):
    """
    Extracts optimization objective information from a cobrapy model.
    Attempts to retrieve coefficients and optimization direction (max/min).
    """
    obj = model.objective
    direction = getattr(obj, "direction", None) or "max"

    obj_coeffs = {}
    expr = getattr(obj, "expression", None)
    if expr is not None:
        try:
            # Get {Reaction/Variable: Coefficient} mapping
            coeffs = expr.as_coefficients_dict()
            for k, v in coeffs.items():
                rid = getattr(k, "id", str(k))
                obj_coeffs[str(rid)] = float(v)
        except Exception:
            pass

    return obj_coeffs, ("max" if str(direction).lower().startswith("max") else "min")

def reaction_equation_safe(r, use_names=False):
    """
    Safely retrieves the reaction equation string from a Reaction object.
    Fallback order:
    1. 'reaction' property
    2. build_reaction_string() method
    3. Manual assembly from stoichiometry (emergency fallback)
    """
    # 1) Try 'reaction' property (available in many cobrapy versions)
    try:
        eq = getattr(r, "reaction", None)
        if isinstance(eq, str) and eq.strip():
            return eq
    except Exception:
        pass

    # 2) Try build_reaction_string()
    try:
        eq = r.build_reaction_string(use_metabolite_names=use_names)
        if isinstance(eq, str) and eq.strip():
            return eq
    except Exception:
        pass

    # 3) Manual assembly (if other methods fail)
    try:
        mets = r.metabolites  # dict: {Metabolite: coefficient}
        left_terms = []
        right_terms = []
        for m, coeff in mets.items():
            name_or_id = m.name if (use_names and getattr(m, "name", None)) else m.id
            sto = abs(coeff)
            term = f"{sto:g} {name_or_id}" if not np.isclose(sto, 1.0) else f"{name_or_id}"
            if coeff < 0:
                left_terms.append(term)
            else:
                right_terms.append(term)
        left = " + ".join(left_terms) if left_terms else "∅"
        right = " + ".join(right_terms) if right_terms else "∅"
        arrow = "<=>" if r.lower_bound < 0 and r.upper_bound > 0 else "->"
        return f"{left} {arrow} {right}"
    except Exception:
        return ""

def export_model_genes_reactions(model, species_name: str, out_dir: Path):
    """
    Exports genes, reactions, and objective information into separate CSV files.
    Generated files:
      - {species}__genes.csv
      - {species}__reactions.csv
      - {species}__objective_reactions.csv
    """
    sp_tag = sanitize_for_filename(species_name)

    # --- Extract Objective Information ---
    obj_coeffs, obj_dir = extract_objective_info(model)
    obj_rxn_ids = set(obj_coeffs.keys())

    # --- Process Gene Data ---
    gene_rows = []
    for g in model.genes:
        gname = getattr(g, "name", "")
        try:
            assoc_rxns = sorted({r.id for r in g.reactions})
        except Exception:
            assoc_rxns = []
        gene_rows.append({
            "gene_id": g.id,
            "gene_name": gname if gname is not None else "",
            "associated_reactions": ";".join(assoc_rxns),
            "reaction_count": len(assoc_rxns),
        })
    genes_df = (pd.DataFrame(gene_rows)
                  .sort_values(["gene_id"], kind="stable")
                  .reset_index(drop=True))

    # --- Process Reaction Data ---
    rxn_rows = []
    for r in model.reactions:
        rname = getattr(r, "name", "")
        subsystem = getattr(r, "subsystem", "")
        lb = getattr(r, "lower_bound", np.nan)
        ub = getattr(r, "upper_bound", np.nan)
        gpr = getattr(r, "gene_reaction_rule", "")

        # 反応式（IDベース／名前ベース）
        eq_ids = reaction_equation_safe(r, use_names=False)
        eq_names = reaction_equation_safe(r, use_names=True)

        coeff = obj_coeffs.get(r.id, 0.0)
        rxn_rows.append({
            "reaction_id": r.id,
            "reaction_name": rname if rname is not None else "",
            "subsystem": subsystem if subsystem is not None else "",
            "lower_bound": float(lb) if lb is not None else np.nan,
            "upper_bound": float(ub) if ub is not None else np.nan,
            "gene_reaction_rule": gpr if gpr is not None else "",
            "objective_coeff": float(coeff),
            "is_in_objective": bool(r.id in obj_rxn_ids),
            "reaction_equation_ids": eq_ids,
            "reaction_equation_names": eq_names,
        })
    reactions_df = (pd.DataFrame(rxn_rows)
                      .sort_values(["is_in_objective", "reaction_id"], ascending=[False, True], kind="stable")
                      .reset_index(drop=True))

    # --- Process Objective Reactions List ---
    obj_rows = [{"reaction_id": rid, "coeff": float(c), "direction": obj_dir}
                for rid, c in obj_coeffs.items()]
    objective_df = (pd.DataFrame(obj_rows)
                      .sort_values(["reaction_id"], kind="stable")
                      .reset_index(drop=True))

    # --- Save Output Files ---
    genes_path = out_dir / f"{sp_tag}__genes.csv"
    rxns_path = out_dir / f"{sp_tag}__reactions.csv"
    obj_path  = out_dir / f"{sp_tag}__objective_reactions.csv"

    genes_df.to_csv(genes_path, index=False)
    reactions_df.to_csv(rxns_path, index=False)
    objective_df.to_csv(obj_path, index=False)

    return genes_path, rxns_path, obj_path

# ========== Export Multiple Species / Models ==========
# Pre-requisite: 'species', 'models', and 'model_dir' must be defined.

for sp, model_file in zip(species, models):
    
    # Construct full path to the model file
    model_path = model_dir / model_file
    
    # Load SBML model
    try:
        model = read_sbml_model(str(model_path))
    except Exception as e:
        print(f"Error reading {model_file}: {e}")
        continue

    # Execute export function
    gp, rp, op = export_model_genes_reactions(model, sp, out_dir)
    print(f"[{sp}] saved:\n  - {gp.name}\n  - {rp.name}\n  - {op.name}")
```

```python
# Extract all reactions involving carbon metabolites (Optimized) ---
# Optimization: Using a list of dictionaries instead of creating DataFrames inside the loop.
# This significantly reduces overhead and execution time.

mtbrxn_all_models = []

for i in tqdm(np.arange(len(models))):
    # Load SBML model
    model = read_sbml_model(model_dir / models[i])
    
    # Use a list to store dictionaries for each row
    data_list = [] 
    for reaction in model.reactions:      
        for metabolite, stoichiometry in reaction.metabolites.items():
            # Append data as a dictionary (much faster than creating a DataFrame)
            data_list.append({
                "reaction": reaction.id,
                "metabolite": metabolite.id,
                "name": metabolite.name,
                "formula": metabolite.formula
            })
    
    # Convert the entire list of dictionaries to a DataFrame at once
    mtbrxn_df = pd.DataFrame(data_list)
    mtbrxn_all_models.append(mtbrxn_df)

# Concatenate results from all models and remove duplicate entries
mtbrxn_all_df = pd.concat(mtbrxn_all_models)
mtbrxn_all_df = mtbrxn_all_df.drop_duplicates(subset=["reaction", "metabolite"])
mtbrxn_all_df = mtbrxn_all_df.reset_index(drop=True)

# Define list of non-carbon elements that might start with 'C'
non_carbons = ["Cl", "Co", "Cu", "Ca", "Cd"]

# Identify presence of carbon atoms based on the molecular formula
mtbrxn_all_df["carbon_atom"] = "Absence"

# Flag metabolites whose formulas start with 'C'
mtbrxn_all_df.loc[mtbrxn_all_df["formula"].str.contains(r'^C', na=False), "carbon_atom"] = "Presence"

# Correct for non-carbon elements
mtbrxn_all_df.loc[mtbrxn_all_df["formula"].isin(non_carbons), "carbon_atom"] = "Absence"

# Filter to retain only carbon-containing metabolites/reactions
mtbrxn_all_df = mtbrxn_all_df[mtbrxn_all_df["carbon_atom"] == "Presence"]
mtbrxn_all_df = mtbrxn_all_df.reset_index(drop=True)

# Display final filtered dataframe
mtbrxn_all_df
```

```python
# Extract a non-redundant list of all metabolites ---
# Optimization: Using a list of dictionaries to aggregate metabolite metadata.
# This prevents the overhead of creating thousands of small DataFrames.
# Estimated execution time: Reduced from ~6 min to a few seconds.

mtb_all_models = []

for i in tqdm(np.arange(len(models))):
    # Load SBML model (using Path object converted to string)
    model = read_sbml_model(str(model_dir / models[i]))
    
    # Store metabolite data in a list of dictionaries for speed
    mtb_data = [] 
    for metabolite in model.metabolites:
        mtb_data.append({
            "metabolite": metabolite.id,
            "name": metabolite.name,
            "formula": metabolite.formula
        })
    
    # Batch-convert to DataFrame and assign species identity
    current_mtb_df = pd.DataFrame(mtb_data)
    current_mtb_df["species"] = species[i]
    mtb_all_models.append(current_mtb_df)

# Consolidate all species data into a single master DataFrame
mtb_df = pd.concat(mtb_all_models)

# Remove duplicates to create a unique metabolite catalog
mtb_df = mtb_df.drop_duplicates(subset=["metabolite"])
mtb_df = mtb_df.reset_index(drop=True)

# List of non-carbon elements to prevent false positive matches
non_carbons = ["Cl", "Co", "Cu", "Ca", "Cd"]

# Default status for carbon atom presence
mtb_df["carbon_atom"] = "Absence"

# Regex check: Flag formulas starting with 'C' as carbon-containing
mtb_df.loc[mtb_df["formula"].str.contains(r'^C', na=False), ["carbon_atom"]] = "Presence"

# Secondary check: Override specific non-carbon elements (e.g., Cl, Ca)
mtb_df.loc[mtb_df["formula"].isin(non_carbons), ["carbon_atom"]] = "Absence"

# Clean up: Remove species column as this is a non-redundant global list
mtb_df = mtb_df.drop(columns="species")

# Final metabolite catalog
mtb_df
```

# Run FBA and extract essential reactions for optimal growth

<div class="alert-success">
<font size="6">the ECOR models (seven representative strains)</font> 
</div>

```python
# ----------------------------
# Parameters and Configuration
# ----------------------------
# Selection of ECOR strains by their numerical IDs
model_sels = [11, 32, 71, 31, 50, 38, 51, 55]

# Threshold for essentiality: KO Growth (KO_Z) < 99.9% of Wild-type Growth (Z)
threshold_factor = 0.999 
eps = 1e-9

# Total Carbon uptake flux limit (mmol C gDW^-1 h^-1) 
# Based on the strategy by Leiby & Marx (2014) to normalize different carbon sources.
C_TOTAL = 120.0  

def _count_carbons_from_formula(formula):
    """
    Counts the number of carbon atoms (nC) from a molecular formula.
    Ensures elements like Cl or Co are excluded using regex.
    """
    if not formula or not isinstance(formula, str):
        return 0
    
    # Regex: Look for 'C' NOT followed by a lowercase letter (to skip Cl, Ca, etc.)
    # Captures the digit following 'C'; if no digit, assumes 1.
    matches = re.findall(r'C(?![a-z])(\d*)', formula)
    if not matches:
        return 0
    
    nC = 0
    for m in matches:
        nC += int(m) if (m and m.isdigit()) else 1
    return nC

def apply_carbon_normalized_medium_from_M9singles(model, carbon_name, M9_singles, aerobic=True, C_TOTAL=120.0):
    """
    Constructs and applies model.medium based on a specific carbon source from the M9_singles data.
    
    Logic:
    - Sets base medium components (salts, vitamins) to their default values (usually 1000).
    - Normalizes the specific carbon source: Upper Bound = C_TOTAL / nC.
    - Returns the number of carbons used and the Exchange reaction ID.
    """
    # Initialize medium template with zeros to ensure a clean slate
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        medium = model.medium.copy()
    for k in list(medium.keys()):
        medium[k] = 0.0

    # Validate if the carbon source exists in the input dataframe columns
    if carbon_name not in M9_singles.columns:
        raise KeyError(f"Column '{carbon_name}' not found in M9_singles.")

    col = M9_singles[carbon_name]

    # Set default uptake limits for all active Exchange (EX) reactions in the TSV
    for ex, val in zip(M9_singles["global_id"], col):
        if model.reactions.has_id(ex):
            v = float(val) if pd.notnull(val) else 0.0
            # cobrapy uses positive values for uptake upper bounds in model.medium
            medium[ex] = max(0.0, v)

    # Apply Carbon Normalization for the target carbon source (type=='Carbon')
    rows = M9_singles[(M9_singles["type"] == "Carbon") & (M9_singles["name"] == carbon_name)]
    nC_used, ex_id_used = 0, None
    for ex, formula in zip(rows["global_id"], rows["formulae"]):
        if model.reactions.has_id(ex):
            nC = _count_carbons_from_formula(formula)
            if nC > 0:
                # Normalize flux: ensures total carbon uptake remains constant across sources
                medium[ex] = C_TOTAL / nC   
                nC_used = nC
                ex_id_used = ex

    # Toggle Oxygen uptake for Aerobic/Anaerobic conditions
    if model.reactions.has_id("EX_o2_e"):
        if not aerobic:
            medium["EX_o2_e"] = 0.0  # Anaerobic condition
        # Aerobic: uses the default value from the TSV (usually 1000)

    # Apply the finalized medium dictionary to the model
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        model.medium = medium

    return nC_used, ex_id_used
```

## Main loop for the ECOR strains

```python
# --- Configuration for ECOR strains ---
# Specific indices for ECOR strains used in the analysis
model_sels = [11, 32, 71, 31, 50, 38, 55]  # ECOR11, ECOR32, ECOR71, ECOR31, ECOR50, ECOR38, ECOR55

# Threshold for essentiality: A reaction is essential if KO growth rate < (threshold_factor * original growth rate)
# Here, a 0.1% decrease in growth triggers essentiality.
threshold_factor = 0.999    
eps = 1e-9

# Total carbon uptake rate (mmol C gDW^-1 h^-1) following Leiby & Marx (2014)
C_TOTAL = 120.0  

# Output control flags
SAVE_ALL_KO = False         # Recommended False if the full knockout result CSV is too large
SAVE_ESSENTIAL_ONLY = True  # Save only the reactions identified as essential
SAVE_SUMMARY = True         # Save a summarized count of essential reactions

def get_growth_status(model):
    """
    Computes the objective value and optimization status of the model.
    
    Args:
        model: COBRApy model object.
    Returns:
        tuple: (Objective value (float or NaN), Status string)
    """
    sol = model.optimize()
    status = getattr(sol, "status", "unknown")
    Z = float(sol.objective_value) if (sol is not None and sol.status == "optimal" and sol.objective_value is not None) else np.nan
    return Z, status

def sanitize_for_filename(s: str) -> str:
    """
    Sanitizes strings to be safe for file paths by replacing non-alphanumeric characters.
    """
    return re.sub(r'[^0-9A-Za-z._-]+', '_', str(s))[:120]

# Setup output directory
out_dir = current_dir.parent / "fba/output/essential_by_species"
out_dir.mkdir(parents=True, exist_ok=True)

# Filter carbon source columns from the M9_singles dataset
# Excludes metadata/annotation columns to isolate experimental conditions
carbon_cols = [c for c in M9_singles.columns if c not in
                ["reaction","metabolite","global_id","flux","keggId","formulae","descriptive_name","type","name","mw"]]

# --- Utility to normalize SRD (Single Reaction Deletion) results ---
def _normalize_srd_result(df: pd.DataFrame) -> pd.DataFrame:
    """
    Normalizes the output of COBRApy's single_reaction_deletion to ensure consistent column names and types.
    Handles fluctuations in index naming and multi-reaction IDs.
    """
    if not isinstance(df, pd.DataFrame):
        df = pd.DataFrame(df)
        
    # Reset index if reaction IDs are stored in the index
    if df.index.name is not None or isinstance(df.index, pd.MultiIndex):
        df = df.reset_index()
        
    # Standardize growth/objective column to 'KO_Z'
    if "growth" in df.columns:
        df = df.rename(columns={"growth": "KO_Z"})
    elif "objective" in df.columns:
        df = df.rename(columns={"objective": "KO_Z"})
        
    # Standardize reaction ID column to 'reaction_id'
    for cand in ["ids", "reaction", "reactions", "index"]:
        if cand in df.columns and "reaction_id" not in df.columns:
            df = df.rename(columns={cand: "reaction_id"})
            
    if "reaction_id" not in df.columns:
        raise KeyError(f"Could not identify reaction_id column from SRD results: {list(df.columns)}")

    # Ensure reaction_id is a string (handles cases where IDs are provided as frozenset/list)
    def _norm_rid(x):
        if isinstance(x, (set, list, tuple)):
            if len(x) == 1:
                return str(next(iter(x)))
            return ";".join(sorted(map(str, x)))  # Join multiple IDs with semicolon
        return str(x)
    
    df["reaction_id"] = df["reaction_id"].apply(_norm_rid)

    # Convert KO_Z to numeric, treating failures (NaN) as lethal (0.0 growth)
    df["KO_Z"] = pd.to_numeric(df["KO_Z"], errors="coerce").fillna(0.0)

    # Normalize status column
    if "status" in df.columns:
        df = df.rename(columns={"status": "status_KO"})
    else:
        df["status_KO"] = "unknown"

    return df[["reaction_id", "KO_Z", "status_KO"]].copy()

# --- Main simulation loop per species ---
for sp, model_file in zip(tqdm(species, desc="Models"), models):
    model_path = model_dir / model_file
    model = read_sbml_model(str(model_path)) 

    # Cache reaction IDs and names for faster lookup
    rxn_ids = [r.id for r in model.reactions]
    rxn_names = {r.id: (r.name if hasattr(r, "name") else "") for r in model.reactions}

    records = []

    # Iterate through each carbon source
    for carbon in tqdm(carbon_cols, desc=f"{sp}", leave=False):
        with model:
            # Apply specific carbon source normalized by C-mmol
            nC_used, ex_id = apply_carbon_normalized_medium_from_M9singles(
                model, carbon, M9_singles, aerobic=True, C_TOTAL=C_TOTAL
            )
            Z, status_base = get_growth_status(model)

            # If the model cannot grow on the baseline condition, record as NA and skip KO analysis
            if not np.isfinite(Z) or Z <= 0:
                records.append(dict(
                    species=sp, carbon=carbon, ex_id=ex_id, nC=nC_used,
                    baseline_Z=np.nan,
                    reaction_id="(NA)", reaction_name="",
                    KO_Z=np.nan, drop_frac=np.nan,
                    essential=False, status_baseline=status_base, status_KO="NA"
                ))
                continue

            # Define the threshold for essentiality based on current baseline growth
            thr = threshold_factor * Z

            # ==== CORE EXECUTION: Batch Single Reaction Deletion ====
            # Note: processes=1 is often faster due to overhead reduction in COBRApy's SRD
            srd_res = single_reaction_deletion(model, rxn_list=rxn_ids, processes=1, method="fba")
            srd_res = _normalize_srd_result(srd_res)

            # Enrich SRD results with metadata
            srd_res["reaction_name"] = srd_res["reaction_id"].map(rxn_names).fillna("")
            srd_res["species"]       = sp
            srd_res["carbon"]        = carbon
            srd_res["ex_id"]         = ex_id
            srd_res["nC"]            = nC_used
            srd_res["baseline_Z"]    = Z
            srd_res["drop_frac"]     = np.where(Z > 0, (Z - srd_res["KO_Z"]) / Z, np.nan)
            srd_res["essential"]     = srd_res["KO_Z"] < thr
            srd_res["status_baseline"] = status_base

            # Accumulate records
            records.extend(srd_res.to_dict(orient="records"))

    # === Data Export per Species ===
    sp_tag = sanitize_for_filename(sp)
    base = f"{sp_tag}"

    result_df = pd.DataFrame.from_records(records)

    # Save full results if requested
    if SAVE_ALL_KO:
        all_path = out_dir / f"{base}__all_ko.csv"
        result_df.to_csv(all_path, index=False)

    # Save only essential reactions
    if SAVE_ESSENTIAL_ONLY:
        essential_df = result_df[result_df["essential"]].copy()
        ess_path = out_dir / f"{base}__essential_only.csv"
        essential_df.to_csv(ess_path, index=False)
        display(essential_df.head())

    # Save summary (count of essential reactions per carbon source)
    if SAVE_SUMMARY:
        essential_df = result_df[result_df["essential"]].copy()
        summary_df = (essential_df
                      .groupby(["species","carbon"])["reaction_id"]
                      .nunique()
                      .reset_index(name="n_essential_reactions"))
        sum_path = out_dir / f"{base}__summary_by_carbon.csv"
        summary_df.to_csv(sum_path, index=False)
        display(summary_df.head())
```

## Measure processing speed

```python
# --- Test speed configuration ---
# Using the same thresholding and carbon uptake logic as the main script
threshold_factor = 0.999    # Reaction is essential if KO_Z < threshold_factor * Z
eps = 1e-9
C_TOTAL = 120.0             # mmol C gDW^-1 h^-1 (Following Leiby & Marx 2014)

# Load the reference model (e.g., iML1515 for E. coli MG1655)
model_path = current_dir.parent / "mg1655/iML1515.xml"
model = read_sbml_model(model_path)

# Prepare carbon source columns and reaction IDs for benchmarking
carbon_cols = [c for c in M9_singles.columns if c not in
                ["reaction","metabolite","global_id","flux","keggId","formulae","descriptive_name","type","name","mw"]]
rxn_ids = [r.id for r in model.reactions]

def KO_growth_once(model, reaction_id):
    """
    Performs a single reaction knockout by setting its bounds to zero 
    and returns the resulting growth rate.
    
    Args:
        model: COBRApy model object.
        reaction_id (str): The ID of the reaction to knock out.
    Returns:
        tuple: (Objective value (float or NaN), Status string)
    """
    try:
        rxn = model.reactions.get_by_id(reaction_id)
    except KeyError:
        # Return NaN if the reaction ID does not exist in the specific model
        return np.nan, "rxn_not_in_model"
    
    with model:
        # Perform knockout by constraining flux to zero
        rxn.lower_bound = 0.0
        rxn.upper_bound = 0.0
        
        try:
            sol = model.optimize()
            # If the optimization fails or is infeasible, assume zero growth
            if sol is None or sol.status != "optimal" or sol.objective_value is None:
                return 0.0, "infeasible"
            
            return float(sol.objective_value), "optimal"
        except Exception:
            # Catch-all for unexpected solver errors
            return 0.0, "error"

def time_seq_loop(model, rxn_ids, Z, thr):
    """
    Benchmark method 1: Sequential loop using KO_growth_once for each reaction.
    This reflects the overhead of repeated manual model adjustments.
    """
    t0 = time.time()
    n_essential = 0
    for rid in rxn_ids:
        KO_Z, status = KO_growth_once(model, rid)

        if status != "optimal":
            continue   # infeasible does not mean essential

        if KO_Z < thr:
            n_essential += 1
            
    return time.time() - t0, n_essential

def time_srd(model, rxn_ids_active, Z, thr, processes=1):
    import time
    t0 = time.time()

    with model:
        res = single_reaction_deletion(
            model,
            rxn_list=rxn_ids_active,
            processes=processes,
            method="fba"
        )

    if "growth" in res.columns:
        res = res.rename(columns={"growth": "KO_Z"})
    elif "objective" in res.columns:
        res = res.rename(columns={"objective": "KO_Z"})

    # ids → reaction_id
    res = res.reset_index(drop=True)

    def _norm(x):
        if isinstance(x, (set, frozenset, list, tuple)):
            return next(iter(x))
        return str(x)

    res["reaction_id"] = res["ids"].apply(_norm)

    #
    res = res[res["reaction_id"].isin(rxn_ids_active)]

    status = res["status"].to_numpy()
    KO_Z   = res["KO_Z"].astype(float).to_numpy()

    n_essential = int(
        ((status == "optimal") & (KO_Z < thr)).sum()
    )

    return time.time() - t0, n_essential

# ---- Set medium and establish baseline growth ----
with model:
    # Use the first available carbon source for the benchmark
    nC_used, ex_id = apply_carbon_normalized_medium_from_M9singles(
        model, carbon_cols[0], M9_singles, aerobic=True, C_TOTAL=C_TOTAL
    )
    Z, status_base = get_growth_status(model)
    
    # Set threshold only if baseline growth is valid
    thr = (threshold_factor * Z) if (Z is not None and np.isfinite(Z) and Z > 0) else 0.0

# baseline solution
sol = model.optimize()
active_rxns = sol.fluxes[sol.fluxes.abs() > 1e-9].index.tolist()

rxn_ids_active = active_rxns   
    
# ---- Performance Benchmarking across 3 approaches ----
# 1. Traditional sequential loop
t_seq, n1 = time_seq_loop(model, rxn_ids_active, Z, thr)

# 2. Optimized batch SRD (Single process)
t_srd1, n2 = time_srd(model, rxn_ids_active, Z, thr, processes=1)

# 3. Optimized batch SRD (Multi-process)
# Adjust 'processes' based on your local machine's CPU core availability
t_srd4, n3 = time_srd(model, rxn_ids_active, Z, thr, processes=4)  

# ---- Summary of Results ----
print(f"--- Benchmark Results (N={len(rxn_ids)} reactions) ---")
print(f"Sequential loop (KO_growth_once x N): {t_seq:7.2f}s, essential counted={n1}")
print(f"SRD (processes=1)                  : {t_srd1:7.2f}s, essential counted={n2}")
print(f"SRD (processes=4)                  : {t_srd4:7.2f}s, essential counted={n3}")
```

## Main

```python
# --- Specific configuration for mg1655 (iML1515) ---
# Update output directory for the single model analysis
out_dir = current_dir.parent / "mg1655/output/essential_by_species"
out_dir.mkdir(parents=True, exist_ok=True)

# Single model setup
sp = "iML1515"  # Species name used for file naming
model_path = current_dir.parent / "mg1655/iML1515.xml"
model = read_sbml_model(model_path)

# Filter carbon source columns (consistent with experimental design)
carbon_cols = [c for c in M9_singles.columns if c not in
                ["reaction","metabolite","global_id","flux","keggId","formulae","descriptive_name","type","name","mw"]]

# Cache reaction IDs and names to avoid repeated attribute access
rxn_ids = [r.id for r in model.reactions]
rxn_names = {r.id: (r.name if hasattr(r, "name") else "") for r in model.reactions}

# Initialize list for species-level results
records = []

for carbon in tqdm(carbon_cols, desc=f"{sp}", leave=False):
    with model:
        # Normalize medium based on carbon molarity
        nC_used, ex_id = apply_carbon_normalized_medium_from_M9singles(
            model, carbon, M9_singles, aerobic=True, C_TOTAL=C_TOTAL
        )
        Z, status_base = get_growth_status(model)

        # Skip knockout analysis if the baseline condition does not support growth
        if not np.isfinite(Z):
            records.append(dict(
                species=sp, carbon=carbon, ex_id=ex_id, nC=nC_used,
                baseline_Z=np.nan,
                reaction_id="(NA)", reaction_name="",
                KO_Z=np.nan, drop_frac=np.nan,
                essential=False, status_baseline=status_base, status_KO="NA"
            ))
            continue

        # Set threshold for essentiality
        thr = threshold_factor * Z  
        
        # Batch knockout simulation
        srd_res = single_reaction_deletion(model, rxn_list=rxn_ids, processes=1, method="fba")

    # --- Data normalization and standardization of SRD results ---
    # Standardize objective/growth column name
    if "growth" in srd_res.columns:
        srd_res = srd_res.rename(columns={"growth":"KO_Z"})
    elif "objective" in srd_res.columns:
        srd_res = srd_res.rename(columns={"objective":"KO_Z"})

    # Ensure reaction IDs are in a column named 'reaction_id'
    srd_res = srd_res.reset_index()  # Handle cases where IDs are in the index
    if "ids" in srd_res.columns:
        srd_res = srd_res.rename(columns={"ids":"reaction_id"})
    elif "index" in srd_res.columns and "reaction_id" not in srd_res.columns:
        srd_res = srd_res.rename(columns={"index":"reaction_id"})

    # Convert growth rates to float and handle missing values as lethal (0.0)
    srd_res["KO_Z"] = srd_res["KO_Z"].astype(float).fillna(0.0)

    # Secondary check to ensure DataFrame structure
    if not isinstance(srd_res, pd.DataFrame):
        srd_res = pd.DataFrame(srd_res)

    if srd_res.index.name is not None or isinstance(srd_res.index, pd.MultiIndex):
        srd_res = srd_res.reset_index()

    # Consolidate possible ID column names into 'reaction_id'
    for cand in ["ids", "reaction", "reactions", "index"]:
        if cand in srd_res.columns and "reaction_id" not in srd_res.columns:
            srd_res = srd_res.rename(columns={cand: "reaction_id"})

    if "reaction_id" not in srd_res.columns:
        raise KeyError("Could not identify 'reaction_id' from SRD results. "
                       "Found columns: {}".format(list(srd_res.columns)))

    # Function to normalize reaction ID types (handles frozensets/lists returned by COBRApy)
    def _norm_reaction_id(x):
        if isinstance(x, (set, list, tuple)):
            if len(x) == 1:
                return str(next(iter(x)))
            return ";".join(sorted(map(str, x)))
        return str(x)

    srd_res["reaction_id"] = srd_res["reaction_id"].apply(_norm_reaction_id)
    
    # --- Enrichment with metadata and analysis flags ---
    srd_res["reaction_name"] = srd_res["reaction_id"].map(rxn_names).fillna("")
    srd_res["species"]       = sp
    srd_res["carbon"]        = carbon
    srd_res["ex_id"]         = ex_id
    srd_res["nC"]            = nC_used
    srd_res["baseline_Z"]    = Z
    srd_res["drop_frac"]     = np.where(Z > 0, (Z - srd_res["KO_Z"]) / Z, np.nan)
    srd_res["essential"]     = srd_res["KO_Z"] < thr
    srd_res["status_baseline"] = status_base
    
    # Standardize optimization status column
    if "status" in srd_res.columns:
        srd_res = srd_res.rename(columns={"status":"status_KO"})
    else:
        srd_res["status_KO"] = "unknown"

    # Append results to the main collection
    records.extend(srd_res.to_dict(orient="records"))

# --- Export and Summary ---
sp_tag = sanitize_for_filename(sp)
base = f"{sp_tag}"

result_df = pd.DataFrame.from_records(records)

# Save the full result set
if SAVE_ALL_KO:
    all_path = out_dir / f"{base}__all_ko.csv"
    result_df.to_csv(all_path, index=False)

# Save the essential reactions only
if SAVE_ESSENTIAL_ONLY:
    essential_df = result_df[result_df["essential"]].copy()
    ess_path = out_dir / f"{base}__essential_only.csv"
    essential_df.to_csv(ess_path, index=False)
    display(essential_df.head())

# Save a summary of essential counts per carbon source
if SAVE_SUMMARY:
    essential_df = result_df[result_df["essential"]].copy()
    summary_df = (essential_df
                  .groupby(["species","carbon"])["reaction_id"]
                  .nunique()
                  .reset_index(name="n_essential_reactions"))
    sum_path = out_dir / f"{base}__summary_by_carbon.csv"
    summary_df.to_csv(sum_path, index=False)
    display(summary_df.head())
```
